1. MAIOR E MENOR ENTRE VARIOS VALORES


```python
primeiro = int(input("Digite um numero: "))
maior = primeiro
menor = primeiro

for i in range (5):
  numero = int(input("Digite um numero: "))
  if numero > maior:
    maior = numero
    if numero < menor:
      menor = numero


print("Maior" , maior)
print("Menor" , menor)
```

4. ESTACIONAMENTO


```python
horas = int(input("Quantas horas ficou estacionado? "))

if horas <= 2:
    valor = 8
elif horas <= 5:
    valor = 8 + (horas - 2) * 3
else:
    valor = 17 + (horas - 5) * 2

print("Valor a pagar: R$", valor)
```

5. MÁQUINA DE VENDA DE INGRESSO

```python
total = 0

while total < 25:
    valor = int(input("Digite o valor inserido: "))
    total += valor

print("Total inserido: R$", total)
print("Troco: R$", total - 25)
```

7. SEGUNDO MAIOR VALOR

```python


maior = None
segundo = None

for i in range(8):
    numero = int(input("Digite um número: "))

    if maior is None or numero > maior:
        segundo = maior
        maior = numero
    elif segundo is None or numero > segundo:
        segundo = numero

print("Maior:", maior)
print("Segundo maior:", segundo)
```

8. DECTECTANDO REPETIÇÃO CONSECUTIVA

```python
anterior = None
repetiu = False

for i in range(8):
    numero = int(input("Digite um número: "))

    if numero == anterior:
        repetiu = True

    anterior = numero

if repetiu:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

9. CONTROLE DE SALDO

```python
saldo = 500

for i in range(6):
    tipo = int(input("Digite o tipo (1 depósito / 2 saque): "))
    valor = float(input("Digite o valor: "))

    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print("Saldo final:", saldo)
```

10. CAIXA ELETRÔNICO SIMPLIFICADO

```python
valor = int(input("Digite o valor: "))

for cedula in [100, 50, 20, 10, 5, 2, 1]:
    quantidade = valor // cedula
    valor = valor % cedula

    print(cedula, ":", quantidade)
```

11. NUMERO PRIMO

```python
numero = int(input("Digite um número: "))

primo = True

for i in range(2, numero):
    if numero % i == 0:
        primo = False
        break

if primo:
    print("É primo")
else:
    print("Não é primo")
```

12. DIVISORES DE UM NÚMERO

```python
numero = int(input("Digite um número positivo: "))

quantidade = 0

for i in range(1, numero + 1):
    if numero % i == 0:
        print(i)
        quantidade += 1

print("Quantidade de divisores:", quantidade)
```

13. NUMEROS PRIMOS EM UM INTERVALO

```python
A = int(input("Digite A: "))
B = int(input("Digite B: "))

quantidade = 0

for numero in range(A, B + 1):
    primo = True

    if numero < 2:
        primo = False

    for i in range(2, numero):
        if numero % i == 0:
            primo = False
            break

    if primo:
        print(numero)
        quantidade += 1

print("Quantidade de primos:", quantidade)
```

14. JOGO COM LIMITES DE TENTATIVAS

```python
secreto = 37

for tentativa in range(1, 6):
    numero = int(input("Digite um número: "))

    if numero == secreto:
        print("Acertou!")
        break
    elif numero < secreto:
        print("O segredo é maior.")
    else:
        print("O segredo é menor.")

    if tentativa == 5:
        print("Você usou as 5 tentativas.")
```
