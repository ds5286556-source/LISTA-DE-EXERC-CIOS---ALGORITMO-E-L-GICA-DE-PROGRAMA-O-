```python
a = int(input("Digite o primeiro numero: "))
b = int(input("Digite o segundo numero: "))

if a > b:
  print("Maior", a)
elif b > a:
  print("Maior", b)
else:
  print("Os dois são iguais")
```

```python

numero = int(input("Digite um número inteiro: "))

if numero % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

if media >= 7:
    print("Aprovado")
elif media >= 5:
    print("Recuperação")
else:
    print("Reprovado")

print("Média:", media)
```

```python
numero1 = int(input("Primeiro número: "))
numero2 = int(input("Segundo número: "))
numero3 = int(input("Terceiro número: "))

maior = numero1

if numero2 > maior:
    maior = numero2

if numero3 > maior:
    maior = numero3

print("Maior número:", maior)
```

```python
n = int(input("Digite N: "))

soma = 0

for i in range(1, n + 1):
    soma = soma + i

print("Soma:", soma)
```

```python
pares = 0

for i in range(10):
    numero = int(input("Digite um número: "))

    if numero % 2 == 0:
        pares = pares + 1

print("Foram digitados", pares, "números pares.")
```
